TASK 1 Resample daily sales into weekly aggregate sales

```python
import matplotlib.pyplot as plt

# Ensure 'Order Date' is in datetime format
df_superstore['Order Date'] = pd.to_datetime(df_superstore['Order Date'])

# Set 'Order Date' as the index
df_superstore_indexed = df_superstore.set_index('Order Date')

# Resample 'Sales' weekly ('W') using sum aggregation
weekly_sales = df_superstore_indexed['Sales'].resample('W').sum()

# Plot the line chart
plt.figure(figsize=(12, 6))
weekly_sales.plot(kind='line')
plt.title('Weekly Sales Trend')
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.grid(True)
plt.show()
```

task 2 calculate and plot 7- day moving average of sales

```python
import matplotlib.pyplot as plt

# Ensure 'Order Date' is in datetime format and set as index (if not already)
df_daily_sales = df_superstore.copy()
df_daily_sales['Order Date'] = pd.to_datetime(df_daily_sales['Order Date'])
df_daily_sales = df_daily_sales.set_index('Order Date')

# Aggregate daily sales
daily_sales = df_daily_sales['Sales'].resample('D').sum()

# Calculate the 7-day rolling mean of daily sales
rolling_mean_7day = daily_sales.rolling(window=7).mean()

# Plot original daily sales and 7-day rolling mean
plt.figure(figsize=(14, 7))
plt.plot(daily_sales, label='Daily Sales', alpha=0.7)
plt.plot(rolling_mean_7day, label='7-Day Rolling Mean', color='red')
plt.title('Daily Sales vs. 7-Day Rolling Mean')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.legend()
plt.grid(True)
plt.show()
```

task 3 Calculate period - over - period daily growth using.shift ()

```python
# Create 'Sales_Prev_Day' using shift(1)
daily_sales_df = daily_sales.to_frame()
daily_sales_df['Sales_Prev_Day'] = daily_sales_df['Sales'].shift(1)

# Calculate daily percentage change in sales
daily_sales_df['Daily_Sales_Change_Pct'] = ((daily_sales_df['Sales'] - daily_sales_df['Sales_Prev_Day']) / daily_sales_df['Sales_Prev_Day']) * 100

print("Daily Sales with previous day's sales and percentage change:")
display(daily_sales_df.head())
```

task 8  Plot distribution  of order profit margin using histplot + KDE

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Calculate Profit Margin, handling cases where Sales might be zero to avoid division by zero errors.
# Filter out rows where 'Sales' is 0 before calculating Profit Margin to prevent NaN/inf values.
# A small epsilon is added to Sales to avoid division by zero errors for infinitesimally small sales, though direct filtering for 0 is usually sufficient.
df_superstore_valid_sales = df_superstore[df_superstore['Sales'] != 0].copy()
df_superstore_valid_sales['Profit_Margin'] = (df_superstore_valid_sales['Profit'] / df_superstore_valid_sales['Sales']) * 100

# Create a histogram with KDE curve for Profit Margin
plt.figure(figsize=(12, 7))
sns.histplot(df_superstore_valid_sales['Profit_Margin'], kde=True, bins=50)
plt.title('Distribution of Profit Margin')
plt.xlabel('Profit Margin (%)')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

task 2  Matplotlib 3D surface plot of synthetic sales - profit surface

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Define the range for the X (Sales) and Y (Profit) axes for the meshgrid.
# These ranges are chosen to be representative of typical values for better visualization
# without being overwhelmed by extreme outliers, and to create a visually interesting smooth surface.
x_surf_range = np.linspace(0, 1000, 50) # Representative Sales range
y_surf_range = np.linspace(-500, 500, 50) # Representative Profit range

X_surf, Y_surf = np.meshgrid(x_surf_range, y_surf_range)

# Define the Z-values (representing Quantity) as a synthetic function of X and Y.
# This function is designed to create a smooth, illustrative surface that could loosely
# represent how Quantity might relate to Sales and Profit.
# For example, Quantity might increase with sales, with some fluctuation or diminishing returns
# based on profit. The specific function Z = (X_surf / 200) + np.sin(Y_surf / 100) + 5
# is chosen for its smoothness and to produce Z values in a reasonable range (e.g., for quantity).
Z_surf = (X_surf / 200) + np.sin(Y_surf / 100) + 5

# Ensure Z values are positive and somewhat realistic for quantity (e.g., minimum of 1).
Z_surf[Z_surf < 1] = 1

# Create a 3D figure
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
# cmap='viridis' provides a good color gradient for visualization.
surface = ax.plot_surface(X_surf, Y_surf, Z_surf, cmap='viridis', edgecolor='none')

# Set labels and title to reflect the user's request
ax.set_xlabel('Sales')
ax.set_ylabel('Profit')
ax.set_zlabel('Quantity')
ax.set_title('Smooth 3D Surface Plot of Hypothetical Quantity vs. Sales and Profit')

# Add a color bar for the Z values (Quantity)
fig.colorbar(surface, shrink=0.5, aspect=10, label='Quantity')

plt.tight_layout()
plt.show()
```

task 6 Reactive dropdown filter (region selector)

```python
import dash
from dash import dcc, html
from dash.dependencies import Input, Output

# Get unique regions from the DataFrame
regions = df_superstore['Region'].unique()

# Initialize the Dash app
app = dash.Dash(__name__)

# Define the app layout
app.layout = html.Div([
    html.H1("Superstore Sales Dashboard"),

    html.Div([
        html.Label("Select Region:"),
        dcc.Dropdown(
            id='region-dropdown',
            options=[{'label': region, 'value': region} for region in regions],
            value=regions[0],  # Default selected value
            multi=True,        # Allow multiple selections
            placeholder="Select one or more regions"
        )
    ], style={'width': '25%', 'display': 'inline-block', 'verticalAlign': 'top', 'padding': '10px'}),

    html.Div(id='output-container', style={'width': '70%', 'display': 'inline-block', 'padding': '10px'})
])

# Define callback to update output based on dropdown selection
@app.callback(
    Output('output-container', 'children'),
    Input('region-dropdown', 'value')
)
def update_output(selected_regions):
    if selected_regions:
        return f"You have selected: {', '.join(selected_regions)}"
    else:
        return "No regions selected."

# Run the app
if __name__ == '__main__':
    app.run(debug=True, mode='inline', port=8051)
```

```python
import pandas as pd
import numpy as np
import datetime
import plotly.express as px

# 1. Generate the mock DataFrame
num_rows = 100
mock_data = {
    'Transaction_ID': [f'TR{i:04d}' for i in range(num_rows)],
    'Category': np.random.choice(['Electronics', 'Office Supplies', 'Furniture', 'Apparel'], num_rows),
    'Region': np.random.choice(['North', 'South', 'East', 'West'], num_rows),
    'Sales': np.random.randint(10, 1000, num_rows) + np.random.rand(num_rows).round(2),
    'Profit': np.round(np.random.normal(loc=50, scale=100, size=num_rows), 2),
    'Date': pd.to_datetime([datetime.date(2023, 1, 1) + datetime.timedelta(days=np.random.randint(0, 365)) for _ in range(num_rows)])
}
mock_df = pd.DataFrame(mock_data)

# Introduce some negative profits to simulate real-world data
negative_profit_indices = np.random.choice(mock_df.index, size=20, replace=False)
mock_df.loc[negative_profit_indices, 'Profit'] = -np.abs(mock_df.loc[negative_profit_indices, 'Profit'])

print("Mock Pandas DataFrame created successfully (first 5 rows):")
display(mock_df.head())

# 2. Aggregate data for the pie chart (Sales by Category)
category_sales = mock_df.groupby('Category')['Sales'].sum().reset_index()

# 3. Create a Plotly Pie Chart
fig_pie = px.pie(category_sales,
                 values='Sales',
                 names='Category',
                 title='Sales Distribution by Category (Mock Data)',
                 hole=0.3, # For a donut chart
                 template='plotly_white' # Clean layout styling
                )

fig_pie.update_traces(textposition='inside', textinfo='percent+label')
fig_pie.update_layout(showlegend=True, title_x=0.5)

print("\nPie chart visualizing sales distribution by category:")
fig_pie.show()
```
